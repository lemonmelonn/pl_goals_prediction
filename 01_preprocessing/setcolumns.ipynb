```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("../preparation/firstclean.csv")
print(df.shape)
```

```python
# Get opponent
df["OPP"] = df["MATCHUP"].str.extract(r"(?:@|vs\.)\s+(.*)")[0].str.strip()

print(df[["MATCHUP", "OPP"]].head())
```

```python
# Map 'W' to 1 and 'L' to 0 (and handle any draws or missing values as NaN or 0 if needed)
df["WIN"] = df["WL"].map({"W": 1, "L": 0})
```

```python
# Home or away game
df["HOME"] = np.where(df["MATCHUP"].str.contains("vs\."), 1, 0)
```

```python
# Save to csv
df.to_csv("../preparation/newcolumns.csv", index=False)
print(df.shape)
```
